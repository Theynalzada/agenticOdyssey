```python
# Importing Dependencies
from langchain.messages import HumanMessage
from pprint import pprint
from pathlib import Path

import warnings
import sys
```

```python
# Filtering out potential warnings
warnings.filterwarnings(action="ignore", 
                        message="Pydantic serializer warnings",
                        category=UserWarning)
```

```python
# Defining the current working directory for the root directory
project_root = Path.cwd().parent

# Appending the path
sys.path.append(str(project_root))

# Importing from Scripts package
from Scripts.travel_planner import build_supervisor_agent, SupervisorContext
```

```python
# Configuring the thread
memory_config={"configurable": {"thread_id": "1"}}

# Calling the function to build the agent
planner_agent=await build_supervisor_agent()

# Constructing a user query
user_query="""
As a citizen of Azerbaijan departing from Baku, 
plan a 5-day trip to Rome, Italy. I am travelling with
my friend and both of us have 1 checked baggage each.
This time we have chosen business as the cabin class.
You can find the dates below.  

Departure date: 2026-10-20 
Return date: 2026-10-25.
"""

# Sending request to the agent
response=await planner_agent.ainvoke(input={"messages": [HumanMessage(content=user_query)]},
                                     config=memory_config,
                                     context=SupervisorContext(user_language="Azerbaijani"))

# Displaying the agent response
response
```

```python
pprint(object=response.get("structured_response").message_to_user)
```
